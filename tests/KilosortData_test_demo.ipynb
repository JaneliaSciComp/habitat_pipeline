```python
# Run integration tests with detailed output
import subprocess
import sys
from pathlib import Path

def run_tests_detailed(category):
    """Run tests with detailed output showing individual test results."""
    
    categories = {
        "loading": "test_kilosort_data_loading.py",
        "analysis": "test_kilosort_data_analysis.py",
        "integration": "test_integration.py"
    }
    
    if category not in categories:
        print(f"Unknown category: {category}")
        return False
    
    test_dir = Path.cwd()  # We're in the tests directory  
    project_root = test_dir.parent
    sys.path.insert(0, str(project_root))
    
    test_file = categories[category]
    test_path = test_dir / test_file
    
    print(f"🧪 Running {category} tests with detailed output")
    print("=" * 50)
    
    # Run pytest with verbose output and capture results to show in notebook
    cmd = ["python", "-m", "pytest", str(test_path), "-v", "--tb=short", "-s"]
    
    try:
        result = subprocess.run(cmd, cwd=str(project_root), capture_output=True, text=True)
        
        # Print the detailed output so it shows in the notebook
        if result.stdout:
            print(result.stdout)
        if result.stderr:
            print("STDERR:", result.stderr)
            
        print(f"\n{'='*50}")
        if result.returncode == 0:
            print(f"🎉 All {category} tests passed!")
        else:
            print(f"💥 Some {category} tests failed.")
        return result.returncode == 0
    except Exception as e:
        print(f"❌ Error running tests: {e}")
        return False

# Run integration tests with details
run_tests_detailed('integration')
```

## 💡 Test Infrastructure Deep Dive

Let's explore how the test infrastructure validates the actual KilosortData implementation:

### 1. Mock Data Creation
Our tests use `create_mock_kilosort_files()` to generate realistic test data that matches the expected Kilosort output format.

```python
import tempfile
import shutil
from pathlib import Path
from conftest import create_mock_kilosort_files
from ingestion.kilosort_data_import import KilosortData

# Create a temporary directory for demonstration
temp_dir = Path(tempfile.mkdtemp())
print(f"📁 Created temp directory: {temp_dir}")

try:
    # Create mock Kilosort data
    print("\n🔧 Creating mock Kilosort data...")
    mock_data = create_mock_kilosort_files(
        temp_dir, 
        include_optional=True, 
        include_cluster_info=True
    )
    
    # Load with actual KilosortData implementation
    print("\n⚡ Loading with actual KilosortData...")
    kilosort_dir = temp_dir / 'kilosort4'
    ks_data = KilosortData(data_input=kilosort_dir)
    
    # Demonstrate key functionality
    print(f"\n📊 KilosortData loaded successfully!")
    print(f"   - Number of clusters: {len(ks_data.ks_ids)}")
    print(f"   - Total spikes: {len(ks_data.spike_times)}")
    print(f"   - Recording duration: {ks_data.duration_seconds:.1f} seconds")
    print(f"   - Animal ID: {ks_data.animal_id}")
    print(f"   - Session ID: {ks_data.session_id}")
    
    # Test firing rates calculation
    firing_rates = ks_data.get_firing_rates()
    print(f"\n🔥 Firing rates calculated for {len(firing_rates)} clusters")
    print(f"   - Rate range: {min(firing_rates.values()):.1f} - {max(firing_rates.values()):.1f} Hz")
    
finally:
    # Cleanup
    shutil.rmtree(temp_dir)
    print(f"\n🗑️ Cleaned up temp directory")
```
